```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from google.colab import drive

drive.mount('/content/drive')


# ===============================
# load image
# ===============================

def load_image(image_path):

    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    if image is None:
        raise FileNotFoundError(f"Unable to load image: {image_path}")

    return image


# ===============================
# bias field generator
# ===============================

def generate_bias_field(shape, coeff=0.5):

    rows, cols = shape

    x = np.linspace(-1, 1, cols)
    y = np.linspace(-1, 1, rows)

    xv, yv = np.meshgrid(x, y)

    bias = 1 + coeff * (xv**2 + yv**2)

    return bias


# ===============================
# apply bias field
# ===============================

def apply_bias_field(image):

    bias = generate_bias_field(image.shape)

    biased_image = image * bias

    biased_image = np.clip(biased_image, 0, 255)

    return biased_image.astype(np.uint8)


# ===============================
# load image
# ===============================

image_path = '/content/drive/My Drive/brain_dataset/sample.png'

image = load_image(image_path)


# ===============================
# augmentation
# ===============================

biased_image = apply_bias_field(image)


# ===============================
# visualization
# ===============================

plt.figure(figsize=(10,5))


plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(biased_image, cmap='gray')
plt.title("Bias Field Augmented")
plt.axis("off")


plt.show()
```
